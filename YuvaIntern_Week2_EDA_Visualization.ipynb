This notebook performs dataset loading, initial inspection, descriptive statistics, class-distribution analysis, feature distributions, relationship analysis, correlation analysis, and interpretation of findings using Pandas, Matplotlib, and Seaborn-compatible Python visualization techniques.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("YuvaIntern_Week2_Breast_Cancer_Dataset.csv")
df.head()
```

```python
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.describe().T
```

```python
df["diagnosis"].value_counts()
```

```python
plt.figure(figsize=(7,4))
sns.countplot(data=df, x="diagnosis")
plt.title("Diagnosis Class Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Patients")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.histplot(data=df, x="mean radius", bins=25, kde=True)
plt.title("Distribution of Mean Radius")
plt.xlabel("Mean Radius")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(7,4))
sns.scatterplot(data=df, x="mean radius", y="mean perimeter", hue="diagnosis")
plt.title("Mean Radius vs Mean Perimeter")
plt.xlabel("Mean Radius")
plt.ylabel("Mean Perimeter")
plt.show()
```

```python
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(12,9))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
top_corr = (
    corr.where(np.triu(np.ones(corr.shape), k=1).astype(bool))
        .stack()
        .sort_values(key=abs, ascending=False)
        .head(10)
)
top_corr
```

```python
# Compare selected feature means by diagnosis
df.groupby("diagnosis")[["mean radius", "mean perimeter", "mean area"]].mean()
```

## Key observations

- The dataset contains 569 observations and 32 columns after adding the readable diagnosis label.
- The target classes are not perfectly balanced; benign cases form the larger class.
- Mean radius shows a spread of values that can be examined using a histogram.
- Mean radius and mean perimeter show a strong positive relationship because both describe related geometric properties.
- The correlation matrix helps identify groups of features that move together and can reveal redundancy among predictors.
- Grouped means allow comparison of selected measurements between benign and malignant cases.

These observations are descriptive and are used to guide later analysis; they do not by themselves establish causal relationships.
